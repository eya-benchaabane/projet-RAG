```python
# ═══════════════════════════════════════════════════════════════════════════
# RAG OPTIMISÉ POUR LA MÉMOIRE - SANS CRASH
# ═══════════════════════════════════════════════════════════════════════════

import fitz
import jsonlines
import faiss
import numpy as np
import os
import re
import sys
import logging
from typing import List, Dict, Tuple, Optional
from dataclasses import dataclass
from datetime import datetime

# ✅ FIX:  Utiliser un modèle plus léger
from sentence_transformers import SentenceTransformer
import requests
import gc

logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - [%(levelname)s] %(message)s',
    handlers=[
        logging. FileHandler('rag.log', encoding='utf-8'),
        logging.StreamHandler()
    ]
)
logger = logging.getLogger(__name__)

# ═══════════════════════════════════════════════════════════════════════════
# CONFIG OPTIMISÉE POUR LA MÉMOIRE
# ═══════════════════════════════════════════════════════════════════════════

@dataclass
class Config:
    PDF_FOLDER: str = 'docs_travel'
    CACHE_DIR: str = 'data/cache'
    
    # ✅ CHUNKING CONSERVATEUR (moins de chunks = moins de mémoire)
    CHUNK_SIZE: int = 300
    CHUNK_OVERLAP: int = 30
    MIN_CHUNK_SIZE: int = 5
    
    # ✅ EMBEDDING MODEL LÉGER (pas BAAI/bge-m3 !)
    # Options: 
    # - 'all-MiniLM-L6-v2' (33MB, bon) ✅ RECOMMANDÉ
    # - 'all-mpnet-base-v2' (438MB, bon mais lourd)
    # - 'BAAI/bge-m3' (1.3GB, trop lourd pour Jupyter) ❌
    EMBEDDING_MODEL: str = 'all-MiniLM-L6-v2'  # ✅ LÉGER
    ENABLE_RERANKING: bool = False  # ✅ DÉSACTIVÉ (économise 1GB)
    
    # Retrieval
    TOP_K_CHUNKS: int = 15  # ✅ RÉDUIT (était 20)
    SIMILARITY_THRESHOLD: float = 0.50
    
    # LLM
    OLLAMA_MODEL: str = 'mistral'
    OLLAMA_URL: str = 'http://localhost:11434'
    OLLAMA_TIMEOUT: int = 300
    TEMPERATURE: float = 0.7
    
    # Storage
    FAISS_FILE: str = 'data/cache/faiss_index.index'
    CHUNKS_FILE: str = 'data/cache/chunks.jsonl'
    
    # ✅ NOUVEAU: Batch processing pour économiser la mémoire
    EMBEDDING_BATCH_SIZE: int = 32  # ✅ RÉDUIT (était 32)
    MAX_CHUNKS_IN_MEMORY: int = 15000  # Si plus de 500 chunks, traiter par batch
    
    def __post_init__(self):
        os.makedirs(self.CACHE_DIR, exist_ok=True)

# ═══════════════════════════════════════════════════════════════════════════
# MEMORY OPTIMIZER
# ═══════════════════════════════════════════════════════════════════════════

class MemoryOptimizer: 
    """Nettoie la mémoire régulièrement"""
    
    @staticmethod
    def cleanup():
        """Force la libération mémoire"""
        gc.collect()
        try:
            import torch
            torch.cuda.empty_cache()
        except: 
            pass
    
    @staticmethod
    def log_memory_usage():
        """Log l'usage mémoire"""
        try:
            import psutil
            process = psutil. Process(os.getpid())
            memory = process.memory_info().rss / 1024 / 1024  # MB
            logger.info(f"💾 Memory usage: {memory:. 1f} MB")
        except:
            pass

# ═══════════════════════════════════════════════════════════════════════════
# PDF EXTRACTOR
# ═══════════════════════════════════════════════════════════════════════════

class PDFExtractor: 
    def __init__(self, config: Config):
        self.config = config
    
    def extract_all_pdfs(self) -> List[Dict]:
        """Extrait les PDFs"""
        
        if not os.path.exists(self.config.PDF_FOLDER):
            raise FileNotFoundError(f"Dossier {self.config.PDF_FOLDER} non trouvé")
        
        pdf_files = [f for f in os.listdir(self.config.PDF_FOLDER)
                    if f.lower().endswith('.pdf')]
        
        if not pdf_files: 
            raise ValueError(f"Aucun PDF dans {self.config.PDF_FOLDER}")
        
        logger. info(f"📄 {len(pdf_files)} PDF(s) à traiter\n")
        
        results = []
        
        for pdf_file in pdf_files:
            path = os.path.join(self.config.PDF_FOLDER, pdf_file)
            logger.info(f"   → {pdf_file}")
            
            try:
                doc = fitz.open(path)
                text = ""
                
                for page in doc:
                    page_text = page.get_text()
                    if page_text. strip():
                        # ✅ Nettoyage rapide
                        page_text = re. sub(r'\s+', ' ', page_text)
                        text += page_text + "\n"
                
                doc.close()
                
                if text.strip():
                    results.append({
                        "filename": pdf_file,
                        "text": text.strip(),
                        "size": len(text)
                    })
                
                logger.info(f"      ✅ {len(text)} chars")
                
                # ✅ Nettoyer la mémoire
                MemoryOptimizer.cleanup()
                
            except Exception as e: 
                logger.error(f"      ❌ {e}")
                continue
        
        logger.info(f"\n✅ {len(results)} PDF(s) traité(s)\n")
        return results

# ═══════════════════════════════════════════════════════════════════════════
# CHUNKER
# ═══════════════════════════════════════════════════════════════════════════

class Chunker:
    def __init__(self, config: Config):
        self.config = config
    
    def chunk_text(self, text: str, filename: str) -> List[Dict]:
        """Découpe le texte"""
        
        # Split simple (pas de regex complexe)
        if '\n\n' in text:
            paragraphs = text.split('\n\n')
        elif '\n' in text:
            paragraphs = text.split('\n')
        else:
            paragraphs = [text]
        
        # Nettoyer
        paragraphs = [p. strip() for p in paragraphs if p.strip() and len(p. split()) > 5]
        
        logger.info(f"✂️ Chunking:  {len(paragraphs)} paras")
        
        chunks = []
        current = ""
        current_words = 0
        
        for para in paragraphs:
            para_words = len(para.split())
            
            if current_words + para_words > self.config.CHUNK_SIZE: 
                if current:
                    chunks.append({
                        "text": current. strip(),
                        "word_count": len(current.split()),
                        "source": filename
                    })
                current = para
                current_words = para_words
            else: 
                current += "\n" + para if current else para
                current_words += para_words
        
        if current:
            chunks.append({
                "text": current.  strip(),
                "word_count": len(current.split()),
                "source": filename
            })
        
        # Filter
        before = len(chunks)
        chunks = [c for c in chunks if c["word_count"] >= self.config.MIN_CHUNK_SIZE]
        
        logger.info(f"   ✅ {len(chunks)} chunks ({before - len(chunks)} supprimés)\n")
        
        return chunks

# ═══════════════════════════════════════════════════════════════════════════
# EMBEDDINGS - OPTIMISÉE
# ═══════════════════════════════════════════════════════════════════════════

class EmbeddingsManager:
    def __init__(self, config: Config):
        self.config = config
        logger.info(f"🧠 Chargement:  {config.EMBEDDING_MODEL}")
        logger.info(f"   (modèle léger, ~33MB)\n")
        
        self.encoder = SentenceTransformer(config.EMBEDDING_MODEL)
        logger.info("✅ Model chargé\n")
    
    def embed_chunks(self, chunks: List[Dict]) -> np.ndarray:
        """Encode les chunks (batch par batch)"""
        
        texts = [c["text"] for c in chunks]
        logger.info(f"📊 Encodage {len(texts)} chunks...")
        
        # ✅ Traiter par batch pour économiser la mémoire
        all_embeddings = []
        batch_size = self.config. EMBEDDING_BATCH_SIZE
        
        for i in range(0, len(texts), batch_size):
            batch = texts[i:i + batch_size]
            logger.info(f"   [{i+1}/{len(texts)}]")
            
            batch_emb = self.encoder.encode(
                batch,
                show_progress_bar=False,
                batch_size=batch_size,
                convert_to_numpy=True
            )
            
            all_embeddings.append(batch_emb)
            
            # ✅ Nettoyer la mémoire après chaque batch
            MemoryOptimizer.cleanup()
        
        embeddings = np.vstack(all_embeddings).astype('float32')
        
        logger.info(f"✅ {len(embeddings)} embeddings créés\n")
        
        return embeddings

# ═══════════════════════════════════════════════════════════════════════════
# FAISS INDEX
# ═══════════════════════════════════════════════════════════════════════════

class FAISSIndex:
    def __init__(self, config: Config):
        self.config = config
        self.index = None
        self. chunks = None
    
    def create(self, embeddings: np.ndarray, chunks: List[Dict]):
        """Crée index"""
        logger.info(f"📁 FAISS:  {len(embeddings)} vecteurs")
        
        dimension = embeddings.shape[1]
        
        # ✅ Utiliser IndexIVFFlat pour économiser la mémoire
        if len(embeddings) > 1000:
            # Pour gros corpus
            nlist = max(100, len(embeddings) // 100)
            quantizer = faiss.IndexFlatL2(dimension)
            self.index = faiss.IndexIVFFlat(quantizer, dimension, nlist)
            self.index.train(embeddings)
            self.index.add(embeddings)
        else:
            # Pour petit corpus
            self.index = faiss.IndexFlatL2(dimension)
            self.index.add(embeddings)
        
        self.chunks = chunks
        logger. info(f"✅ Index créé\n")
    
    def save(self):
        """Sauvegarde"""
        os.makedirs(os.path.dirname(self.config.FAISS_FILE), exist_ok=True)
        faiss.write_index(self.index, self.config.FAISS_FILE)
        logger.info("💾 Index sauvegardé\n")
    
    def load(self) -> bool:
        """Charge"""
        if not os.path.exists(self.config.FAISS_FILE):
            return False
        
        try: 
            self.index = faiss. read_index(self.config. FAISS_FILE)
            logger.info(f"⏫ Index chargé: {self.index.ntotal} vecteurs\n")
            return True
        except Exception as e:
            logger.error(f"❌ Erreur:  {e}\n")
            return False
    
    def search(self, query_emb: np.ndarray, k: int) -> Tuple[List[int], List[float]]:
        """Recherche"""
        distances, indices = self.index.search(query_emb, min(k, self.index.ntotal))
        return indices[0]. tolist(), distances[0].tolist()

# ═══════════════════════════════════════════════════════════════════════════
# RETRIEVER
# ═══════════════════════════════════════════════════════════════════════════

class Retriever:
    def __init__(self, config: Config, embeddings: EmbeddingsManager, index: FAISSIndex):
        self.config = config
        self.embeddings = embeddings
        self.index = index
    
    def retrieve(self, query: str) -> List[Dict]:
        """Retrieval"""
        logger.info(f"🔍 Retrieval: '{query}'")
        
        query_emb = self.embeddings.encoder.encode([query])
        query_emb = np.array(query_emb).astype('float32')
        
        k = min(self.config.TOP_K_CHUNKS * 2, len(self.index.chunks))
        indices, distances = self.index.search(query_emb, k)
        
        results = []
        for idx, dist in zip(indices, distances):
            chunk = self.index.chunks[idx]. copy()
            chunk["score"] = 1 / (1 + dist)
            results.append(chunk)
        
        filtered = [c for c in results
                   if c["score"] > self.config.SIMILARITY_THRESHOLD]
        
        logger. info(f"✅ {len(filtered)} chunks trouvés\n")
        
        return filtered[: self.config.TOP_K_CHUNKS]

# ═══════════════════════════════════════════════════════════════════════════
# OLLAMA LLM
# ═══════════════════════════════════════════════════════════════════════════

class OllamaLLM:
    def __init__(self, config: Config):
        self.config = config
        self._check()
    
    def _check(self) -> bool:
        try:
            r = requests.get(f"{self.config.OLLAMA_URL}/api/tags", timeout=5)
            if r.status_code == 200:
                models = [m['name'] for m in r.json().get('models', [])]
                logger.info(f"✅ Ollama:  {models}\n")
                return True
        except Exception as e:
            logger.error(f"❌ Ollama: {e}\n")
            return False
    
    def generate(self, prompt: str) -> str:
        try:
            r = requests.post(
                f"{self.config. OLLAMA_URL}/api/generate",
                json={
                    "model": self.config.OLLAMA_MODEL,
                    "prompt": prompt,
                    "stream": False,
                    "options": {"temperature": self.config.TEMPERATURE}
                },
                timeout=self.config.OLLAMA_TIMEOUT
            )
            
            if r.status_code == 200:
                return r.json()['response']. strip()
            return f"Erreur: {r.status_code}"
        except Exception as e:
            return f"Erreur: {str(e)}"

# ═══════════════════════════════════════════════════════════════════════════
# MAIN RAG
# ═══════════════════════════════════════════════════════════════════════════

class RAGSystem:
    def __init__(self, config: Config = None):
        self.config = config or Config()
        
        self.pdf_extractor = PDFExtractor(self.config)
        self.chunker = Chunker(self. config)
        self.embeddings = EmbeddingsManager(self. config)
        self.faiss_index = FAISSIndex(self.config)
        self.retriever = Retriever(self.config, self.embeddings, self.faiss_index)
        self.llm = OllamaLLM(self.config)
        
        self.initialized = False
        logger.info("🚀 RAG créé (optimisé mémoire)\n")
    
    def initialize(self, force_rebuild: bool = False) -> bool:
        """Initialize"""
        logger.info("="*80)
        logger.info("[🚀 INIT]")
        logger.info("="*80 + "\n")
        
        try:
            if not force_rebuild and self.faiss_index.load():
                if os.path.exists(self.config.CHUNKS_FILE):
                    with jsonlines.open(self.config. CHUNKS_FILE) as reader:
                        self.faiss_index.chunks = list(reader)
                    logger.info(f"✅ {len(self.faiss_index.chunks)} chunks chargés\n")
                    self.initialized = True
                    return True
            
            # BUILD
            logger.info("[1/4] EXTRACTION\n")
            pdfs = self.pdf_extractor.extract_all_pdfs()
            
            logger.info("[2/4] CHUNKING\n")
            all_chunks = []
            for pdf in pdfs: 
                chunks = self.chunker.chunk_text(pdf["text"], pdf["filename"])
                all_chunks.extend(chunks)
            
            logger.info(f"✅ Total: {len(all_chunks)} chunks\n")
            
            logger.info("[3/4] EMBEDDINGS\n")
            embeddings = self.embeddings.embed_chunks(all_chunks)
            
            logger.info("[4/4] FAISS\n")
            self.faiss_index.create(embeddings, all_chunks)
            self.faiss_index.save()
            
            os.makedirs(self.config. CACHE_DIR, exist_ok=True)
            with jsonlines.open(self.config. CHUNKS_FILE, mode="w") as writer:
                for chunk in all_chunks:
                    writer.write(chunk)
            
            self.initialized = True
            logger. info("✅ READY\n")
            return True
        
        except Exception as e: 
            logger.error(f"❌ {e}\n")
            import traceback
            traceback.print_exc()
            return False
    
    def ask(self, question: str) -> Dict:
        """Ask"""
        if not self.initialized:
            return {"error": "RAG non init"}
        
        logger.info(f"[? ] {question}\n")
        
        chunks = self.retriever.retrieve(question)
        
        if not chunks:
            return {
                "answer": "Pas d'information",
                "chunks": 0
            }
        
        context = "\n\n".join([f"[{c['source']}]\n{c['text']}"
                              for i, c in enumerate(chunks, 1)])
        
        prompt = f"""Tu es un assistant expert qui répond aux questions en utilisant exclusivement les extraits de documents fournis ci-dessous.
Analyse soigneusement le contexte, puis rédige une réponse claire, structurée et précise à la question posée.
Si aucune information pertinente n’est présente dans le contexte, indique simplement : "Je ne sais pas." : 
        
CONTEXTE:
{context}

QUESTION:  {question}

RÉPONSE: """
        
        answer = self.llm.generate(prompt)
        
        logger.info("✅ Done\n")
        
        return {
            "answer":  answer,
            "chunks": len(chunks)
        }

# ═══════════════════════════════════════════════════════════════════════════
# USAGE
# ═══════════════════════════════════════════════════════════════════════════

if __name__ == "__main__": 
    print("\n🚀 RAG OPTIMISÉ MÉMOIRE\n")
    
    config = Config()
    print(f"Config: {config. EMBEDDING_MODEL} (léger)")
    print(f"Batch size: {config.EMBEDDING_BATCH_SIZE}\n")
    
    rag = RAGSystem(config)
    
    if rag.initialize(force_rebuild=True):
        result = rag.ask("parle moi des Musée archéologique en Tunisie")
        print(f"\n✅ {result['chunks']} chunks")
        print(f"Réponse: {result['answer'][: 400]}...")
    else:
        print("❌ Erreur")
```

```python
import os, json, uuid

class ChatHistoryManager:
    def __init__(self, base_dir="data/chats"):
        os.makedirs(base_dir, exist_ok=True)
        self.base_dir = base_dir
        self.current_chat_id = None

    def new_chat(self):
        chat_id = f"chat_{uuid.uuid4().hex[:8]}"
        open(os.path.join(self.base_dir, chat_id + ".jsonl"), "w").close()
        self.current_chat_id = chat_id
        return chat_id

    def save_message(self, role, content):
        if not self.current_chat_id:
            self.new_chat()
        path = os.path.join(self.base_dir, self.current_chat_id + ".jsonl")
        with open(path, "a", encoding="utf-8") as f:
            f.write(json.dumps({"role": role, "content": content}, ensure_ascii=False) + "\n")

    def load_history(self, chat_id=None):
        if chat_id:
            self.current_chat_id = chat_id
        if not self.current_chat_id:
            return []
        path = os.path.join(self.base_dir, self.current_chat_id + ".jsonl")
        history = []
        if os.path.exists(path):
            with open(path, "r", encoding="utf-8") as f:
                for line in f:
                    msg = json.loads(line)
                    history.append((msg["role"], msg["content"]))
        return history

    def list_chats(self):
        return sorted([f.replace(".jsonl","") for f in os.listdir(self.base_dir)], reverse=True)
```

```python
import gradio as gr
from typing import List
import logging

logger = logging.getLogger(__name__)

class ChatbotInterface:
    """Interface Gradio pour le RAG - initialisation automatique safe"""
   
    def __init__(self, rag_system):
        self.rag = rag_system
        self.initializing = False  # Flag pour savoir si l'init est en cours
   
    def _initialize_rag(self):
        """Initialise le RAG si nécessaire"""
        if not self.rag.initialized and not self.initializing:
            self.initializing = True
            logger.info("⏳ Initialisation automatique du RAG...")
            success = self.rag.initialize(force_rebuild=False)
            if success:
                logger.info("✅ RAG prêt !")
            else:
                logger.error("❌ L'initialisation du RAG a échoué")
            self.initializing = False
   
    def chat(self, message: str, history: List) -> str:
        """Traite un message - avec vérification de l'initialisation"""
        if not message.strip():
            return "⚠️ Veuillez entrer une question"
       
        # Commande spéciale pour réinitialiser le contexte
        if message.lower() in ["nouveau sujet", "reset", "réinitialiser"]:
            self.rag.reset_context()
            return "✅ Contexte réinitialisé. Vous pouvez commencer un nouveau sujet!"
       
        # Initialisation automatique si nécessaire
        if not self.rag.initialized:
            self._initialize_rag()
            if not self.rag.initialized:
                return "⏳ RAG en cours de chargement, veuillez patienter..."
       
        try:
            logger.info(f"📝 Question: {message}")
            result = self.rag.ask(message)
            answer = result.get('answer', 'Erreur de génération')
            return answer
        except Exception as e:  
            logger.error(f"❌ Erreur:  {str(e)}")
            return f"❌ Erreur: {str(e)}"
   
    def build_interface(self) -> gr.Blocks:
        """Construit l'interface avec ChatInterface"""
       
        with gr.Blocks(title="🌍 Chatbot Voyage RAG") as demo:
           
            gr.HTML("""
            <div style='background:  linear-gradient(135deg, #667eea 0%, #764ba2 100%);
                        color: white; padding: 30px; border-radius: 15px; text-align: center;'>
                <h1>🌍 Chatbot Voyage Intelligent</h1>
                <p>Posez vos questions sur les destinations de voyage</p>
                <p style='font-size: 0.9em; opacity: 0.9;'>💡 Tapez "nouveau sujet" pour réinitialiser le contexte</p>
            </div>
            """)
           
            gr.Markdown("### 💬 Chat avec RAG - Mémoire Contextuelle Activée")
           
            gr.ChatInterface(
                fn=self.chat,
                examples=[
                    "Parle moi de la Tunisie",
                    "Quel est son climat?",
                    "Où aller en tunisie pour voir la nature?",
                    "Quels sont les meilleurs plages en tunisie?",
                    "comment visiter carthage?"
                ]
            )
           
            gr.HTML("""
            <div style='text-align: center; border-top: 1px solid #ddd; padding-top: 20px; color: #666;'>
                <p>🤖 RAG Professionnel avec Mémoire Contextuelle | 📚 Base PDF | ⚡ Ollama Gemma2</p>
                <p style='font-size: 0.85em;'>Commandes spéciales: "nouveau sujet" ou "reset" pour réinitialiser le contexte</p>
            </div>
            """)
       
        return demo
   
    def launch(self, share=False, server_port=7876):
        """Lance l'interface"""
        demo = self.build_interface()
        logger.info(f"🚀 Lancement sur http://localhost:{server_port}")
        demo.launch(
            share=share,
            server_port=server_port,
            show_error=True
        )

# ═══════════════════════════════════════════════════════════════════════════
# LANCEMENT
# ═══════════════════════════════════════════════════════════════════════════


print("\n" + "="*80)
print("🎨 Création de l'interface Gradio...")
print("="*80 + "\n")

chatbot_ui = ChatbotInterface(rag)
chatbot_ui.launch(share=False)
```

```python
import logging

# NIVEAU DE LOG OPTIMAL POUR DEBUG
logging.basicConfig(level=logging.INFO)

config = Config()
rag = RAGSystem(config)

def debug_ask(question, force_rebuild=False):
    print("=== [DEBUG RAG] ===")
    print("Chemin PDF:", config.PDF_FOLDER)
    print("Modèle d'embedding:", config.EMBEDDING_MODEL)
    
    # Vérification des PDFs
    try:
        pdf_files = os.listdir(config.PDF_FOLDER)
        pdfs = [f for f in pdf_files if f.lower().endswith('.pdf')]
        print(f"PDFs détectés dans le dossier : {pdfs if pdfs else '-- Aucun --'}")
    except Exception as e:
        print(f"Erreur accès PDF folder : {e}")
        return
    
    # Initialisation RAG
    if rag.initialize(force_rebuild=force_rebuild):
        print("✅ Index et chunks initialisés !")
        print(f"Nombre total de chunks : {len(rag.faiss_index.chunks) if rag.faiss_index.chunks else 0}")
        
        chunks_retrieved = rag.retriever.retrieve(question)
        print(f"\nChunks récupérés pour la question : '{question}'")
        print(f"Nombre de chunks récupérés : {len(chunks_retrieved)}")
        
        for i, chunk in enumerate(chunks_retrieved):
            print(f"\n--- Chunk {i} : score {chunk.get('score', 'N/A'):.3f} ---")
            print(chunk['text'][:250].replace('\n',' '))
        
        # Envoi au LLM
        print("\n--- Réponse Ollama ---")
        result = rag.ask(question)
        print(result['answer'])
        print(f"\nTotal chunks utilisés dans la réponse : {result['chunks']}")
    else:
        print("❌ Échec init RAG (chunks ou index manquants)")

# ➡️ Pose ta question ici :
debug_ask("Parler moi de l'histoire de la Tunisie", force_rebuild=False)
```
